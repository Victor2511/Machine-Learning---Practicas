```python
df = dm_to_df(dd1, data["user_id"])
df
```

```python
Z = []
```

```python
Z.append([1,10,0.7,2])#id1, id2, d, n_elementos_en_cluster -> 11.
```

```python
# Obtenemos las columnas actuales antes de agregar la 11
columnas_actuales = df.columns.values.tolist()

for i in columnas_actuales:
    # Usar sintaxis .loc[fila, columna] sin corchetes dobles
    dist_min = min(df.loc[1, i], df.loc[10, i])
    df.loc[11, i] = dist_min
    df.loc[i, 11] = dist_min

# La distancia del clúster 11 consigo mismo debe ser 0.0
df.loc[11, 11] = 0.0

df
```

```python
def agrupar_clusters(df, x, y, n, Z):
    # 1. Guardar en Z: [id1, id2, distancia entre ellos, num_elementos]
    # Usamos loc[x, y] para obtener la distancia real antes de eliminar
    distancia = df.loc[x, y]
    Z.append([x, y, distancia, 2]) # Ajusta el último número si es un clúster con más elementos
    
    # 2. Calcular distancias mínimas para el nuevo clúster 'n'
    columnas_actuales = df.columns.values.tolist()
    for i in columnas_actuales:
        dist_min = min(df.loc[x, i], df.loc[y, i])
        df.loc[n, i] = dist_min
        df.loc[i, n] = dist_min
        
    # Diagonal principal en cero
    df.loc[n, n] = 0.0
    
    # 3. Eliminar los clústeres viejos (x e y) de filas y columnas
    df = df.drop(index=[x, y], columns=[x, y])
    
    return df
```

```python
Z = [] # Inicializar lista Z

# Agrupar 1 y 10 -> Clúster 11
df = agrupar_clusters(df, x=1, y=10, n=11, Z=Z)
df
```

```python
# Agrupar 2 y 7 -> Clúster 12
df = agrupar_clusters(df, x=2, y=7, n=12, Z=Z)
df
```

```python
# Agrupar 5 y 8 -> Clúster 13
df = agrupar_clusters(df, x=5, y=8, n=13, Z=Z)
df
```

```python
df = agrupar_clusters(df, x=11, y=13, n=14, Z=Z)
df
```

```python
def agrupar_clusters3(df, x, y, z, n, Z):
    # 1. Calcular la distancia entre los elementos agrupados (el mínimo entre los pares)
    distancia = min(df.loc[x, y], df.loc[x, z], df.loc[y, z])
    
    # Guardar en Z: [id1, id2, dist, n_elementos]
    # Nota: Scipy/Matplotlib en Z solo guarda 2 elementos agrupados en el paso actual y la suma de sus miembros
    Z.append([x, y, distancia, 3]) 
    
    # 2. Calcular distancias mínimas entre el nuevo clúster 'n' y los demás
    columnas_actuales = df.columns.values.tolist()
    for i in columnas_actuales:
        dist_min = min(df.loc[x, i], df.loc[y, i], df.loc[z, i])
        df.loc[n, i] = dist_min
        df.loc[i, n] = dist_min
        
    # Diagonal principal en cero
    df.loc[n, n] = 0.0
    
    # 3. Eliminar los clústeres viejos (x, y, z) de filas y columnas
    df = df.drop(index=[x, y, z], columns=[x, y, z])
    
    return df
```

```python
df = agrupar_clusters3(df, x=9, y=12, z=14, n=15, Z=Z)
df
```

```python
df = agrupar_clusters3(df, x=4, y=6, z=15, n=16, Z=Z)
df
```

```python
df = agrupar_clusters(df, x=3, y=16, n=17, Z=Z)
df
```

# Clustering jerarquico

```python
import matplotlib.pyplot as plt
from scipy.cluster.hierarchy import dendrogram, linkage
```

```python
movies
```

```python
data[movies]
```

```python
Z = linkage(data[movies], "ward")
Z
plt.figure(figsize=(25,10))
plt.title("Dendrograma jerarquico para el clustering")
plt.xlabel("ID de los usuarios de Netflix")
plt.ylabel("Distancia")
dendrogram(Z, leaf_rotation=90., leaf_font_size=10.0)
plt.show()
```

```python
Z = linkage(data[movies], "average")
Z
plt.figure(figsize=(25,10))
plt.title("Dendrograma jerarquico para el clustering")
plt.xlabel("ID de los usuarios de Netflix")
plt.ylabel("Distancia")
dendrogram(Z, leaf_rotation=90., leaf_font_size=10.0)
plt.show()
```
